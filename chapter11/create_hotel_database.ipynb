```python
import sqlite3
import requests

url = "https://raw.githubusercontent.com/roberto-inf/building-llm-applications/main/ch11/hotel_db/cornwall_hotels_schema.sql"
db_filename = "cornwall_hotels.db"

try:
    # 1. Fetch the SQL schema directly into memory
    print("Fetching SQL schema from GitHub...")
    response = requests.get(url)
    response.raise_for_status()
    sql_script = response.text

    # 2. Connect to the SQLite database (creates the file if it doesn't exist)
    print(f"Creating and populating {db_filename}...")
    with sqlite3.connect(db_filename) as conn:
        cursor = conn.cursor()
        
        # 3. Execute the entire SQL script at once
        cursor.executescript(sql_script)
        
        # 4. Commit the changes
        conn.commit()

    print("Success! Database created and schema initialized.")

except requests.exceptions.RequestException as req_err:
    print(f"Failed to download the schema: {req_err}")
except sqlite3.Error as sql_err:
    print(f"SQLite error occurred: {sql_err}")
```
